```python
def mediana(lista_ord):
  soma = 0
  cont = 0
  for i in range(len(lista_ord)):
    if i == len(lista_ord) // 2 or (len(lista_ord) % 2 == 0 and i == len(lista_ord) // 2 - 1):
      soma += lista_ord[i]
      cont += 1
  return soma / cont
```

4. Use a notação Big O para descrever a complexidade da função abaixo, justificando sua resposta. A função retorna True se a lista for uma lista “soma-100”, e False caso contrário. Uma lista “soma-100” atende o seguinte critério: a soma do 1º número com o último dá 100, a soma do 2º com o penúltimo da 100, a soma do 3º com o antepenúltimo dá 100, e assim por diante.

```python
def soma_100(lista):
  esq = 0
  dir = len(lista) - 1
  while esq < len(lista) // 2:
    if lista[esq] + lista[dir] != 100:
      return False
    esq += 1
    dir -= 1
  return True
```

5. Use a notação Big O para descrever a complexidade da função abaixo, justificando sua resposta. A função encontra e retorna o maior produto de 3 números de uma lista dada.

```python
def maior_produto(lista):
  maior = lista[0] * lista[1] * lista[2]
  for i in range(len(lista)):
    for j in range(i + 1, len(lista)):
      for k in range(j + 1, len(lista)):
        if lista[i] * lista[j] * lista[k] > maior:
          maior = lista[i] * lista[j] * lista[k]
  return maior
```

9. O código abaixo define uma função que verifica se um array de números contém um par de números cuja soma é 10. Esta implementação da função tem complexidade O(N2). Otimize o código de modo que a função realize a mesma tarefa, mas com complexidade O(N).

```python
def soma_10(array):
  for i in range(len(array)):
    for j in range(len(array)):
      if i != j and array[i] + array[j] == 10:
        return True
  return False
```
