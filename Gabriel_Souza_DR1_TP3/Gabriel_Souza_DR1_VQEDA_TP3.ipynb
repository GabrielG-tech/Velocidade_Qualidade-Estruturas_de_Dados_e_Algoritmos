1. O código abaixo define uma função recursiva que recebe 2 números inteiros em parâmetros chamados “inicio” e “fim”. A função imprime número sim, número não, na faixa de “inicio” até “fim”. Por exemplo, se “inicio” for 0 e “fim” for 10, a função deve imprimir 0, 2, 4, 6, 8 e 10. Identifique o caso base dessa recursão, justificando sua resposta.

```python
def imprime_pulando(inicio, fim):
  if inicio <= fim:
    print(inicio)
    imprime_pulando(inicio + 2, fim)
# imprime_pulando(0, 10)
```

**Resposta**: A função imprime_pulando imprime números de inicio até fim, pulando de dois em dois.
O caso base ocorre quando o "inicio" ultrapassa o "fim", pois a função deve parar de chamar a si mesma. Isso evita uma recursão infinita. O critério é estabelecido na condição "if inicio <= fim".

2. O código abaixo define uma função que recebe 2 números inteiros em parâmetros chamados “inicio” e “fim”. A função deveria retornar a soma de todos os números inteiros na faixa de “inicio” até “fim”. Por exemplo, se “inicio” for 1 e “fim” for 10, a função deve retornar a soma de todos os números inteiros de 1 até 10, que é 55. Entretanto, está faltando o caso base no código, e ele executará indefinidamente! Corrija o código adicionando o caso base correto.

```python
def soma(inicio, fim):
  return fim + soma(inicio, fim - 1)
```

**Resposta**: A função soma está incompleta porque falta um **caso base** para impedir chamadas infinitas.

```python
def soma(inicio, fim):
    if inicio > fim:  # Caso base: quando o "inicio" ultrapassar o "fim"
        return 0
    return fim + soma(inicio, fim - 1)
soma(2, 5)
```

3. Escreva uma função recursiva que recebe um array de strings e retorna o nº total de caracteres em todas as strings. Por exemplo, se o argumento passado para a função for o array ["ab", "c", "def", "ghij"], ela deve retornar 10, pois há 10 caracteres no total.

```python
def contar_caracteres(lista):
    if not lista:  # Caso base: se a lista for vazia
        return 0
    return len(lista[0]) + contar_caracteres(lista[1:])
contar_caracteres(["ab", "c", "def", "ghij"])
```

4. Escreva uma função recursiva que recebe um array de números inteiros e retorna um novo array contendo apenas os números pares.

```python
def filtrar_pares(lista):
    if not lista:
        return []
    primeiro, resto = lista[0], lista[1:]
    return ([primeiro] if primeiro % 2 == 0 else []) + filtrar_pares(resto)
filtrar_pares([1, 2, 3, 4, 5, 6])
```

5. Escreva uma função recursiva que recebe uma string e retorna o 1º índice que contém o caractere ‘x’. Por exemplo, na string "abcdefghijklmnopqrstuvwxyz", o 1º (e único) ‘x’ está no índice 23. Caso a string não contenha o caractere ‘x’, a função deve retornar -1.

```python
def encontrar_x(s, indice=0):
    if indice >= len(s):
        return -1
    return indice if s[indice] == 'x' else encontrar_x(s, indice + 1)
print(encontrar_x("abcdefghijklmnopqrstuvwxyz"))
print(encontrar_x("abcdef"))

```

6. Escreva um programa que receba como argumento de linha de comando uma string e imprima o caminho de todos os arquivos existentes no diretório corrente (incluindo os arquivos dos subdiretórios) cujos nomes incluam a string de entrada.

```python
import os

def buscar_arquivos(string, caminho="."):
    arquivos_encontrados = []
    for raiz, _, arquivos in os.walk(caminho): # percorre todos os dirs e subdirs a partir do dir atual "caminho"
        for arquivo in arquivos:
            if string in arquivo:
                arquivos_encontrados.append(os.path.join(raiz, arquivo))
    return arquivos_encontrados

# Exemplo de uso com print
resultado = buscar_arquivos("img")
for arquivo in resultado:
    print(arquivo)
```

7. Escreva uma função recursiva chamada “potencia” para elevar um nº a uma potência. Ela deverá ter 2 parâmetros: base (int ou float) e expoente (int). Teste sua função com várias combinações, incluindo expoentes positivos e negativos, além dos casos especiais em que o expoente é 0 ou 1 (x0 = 1 e x1 = x).
Dica: para lidar com expoentes negativos, você pode usar a transformação xy = 1/(x-y).

```python
def potencia(base: float, expoente: int):
    if expoente == 0:
        return 1
    if expoente < 0:
        return 1 / potencia(base, -expoente)
    return base * potencia(base, expoente - 1)

print(potencia(2, 3))  # Saída: 8
print(potencia(2, -2))  # Saída: 0.25
```

9. O código abaixo define uma função recursiva que recebe uma lista de números e um índice inicial. A função retorna a soma de todos os números da lista a partir do índice inicial, desde que um nº em particular não faça a soma ultrapassar 100. Se somar um nº em particular fizer a soma ultrapassar 100, este nº será ignorado. Entretanto, ela faz chamadas recursivas desnecessárias. Corrija o código para eliminar as chamadas desnecessárias.

```python
def soma_ate_100(lista, inicio=0):
  if len(lista) - inicio == 0:
    return 0
  if lista[inicio] + soma_ate_100(lista, inicio + 1) > 100:
    return soma_ate_100(lista, inicio + 1)
  return lista[inicio] + soma_ate_100(lista, inicio + 1)
```

**Resposta**: A função percorre a lista uma vez, logo, tem complexidade → **O(N)**.

```python
def soma_ate_100(lista, inicio=0, soma=0):
    if inicio >= len(lista):
        return soma
    if soma + lista[inicio] > 100:
        return soma_ate_100(lista, inicio + 1, soma)
    return soma_ate_100(lista, inicio + 1, soma + lista[inicio])

print(soma_ate_100([5, 5, 5, 5, 20, 10, 10, 50, 10]))  
print(soma_ate_100([5, 5, 10]))  
print(soma_ate_100([101, 102, 100, 5]))  
```

11. O código abaixo define uma função recursiva que calcula o N-ésimo número de uma sequência matemática conhecida como sequência de Golomb. Entretanto, ela é terrivelmente ineficiente! Reescreva a função usando memoização para otimizá-la.

```python
def golomb(n):
  if n == 1:
    return 1
  return 1 + golomb(n - golomb(golomb(n - 1)))
print(golomb(1))
print(golomb(2))
print(golomb(3))
print(golomb(4))
print(golomb(5))
print(golomb(6))
```

Por exemplo, a figura abaixo mostra um grid com 3 linhas e 7 colunas, onde queremos ir da célula indicada por “S” à célula indicada por “F”.



A cada passo, podemos nos mover uma posição para a direita:



Ou uma posição para baixo:



O código abaixo define uma função recursiva que recebe o nº de linhas e o nº de colunas do grid e calcula o total de caminhos mais curtos entre as células “S” e “F”. Reescreva a função usando memoização para melhorar sua eficiência.



```python
def caminhos_unicos(linhas, colunas):
  if linhas == 1 or colunas == 1:
    return 1
  return caminhos_unicos(linhas - 1, colunas) + caminhos_unicos(linhas, colunas - 1)
```

```python
def caminhos_unicos(linhas, colunas, memo={}):
    if (linhas, colunas) in memo:
        return memo[(linhas, colunas)]
    if linhas == 1 or colunas == 1:
        return 1
    memo[(linhas, colunas)] = caminhos_unicos(linhas - 1, colunas, memo) + caminhos_unicos(linhas, colunas - 1, memo)
    return memo[(linhas, colunas)]

linhas = 3
colunas = 7
resultado = caminhos_unicos(linhas, colunas)
print(f'O número de caminhos mais curtos em um grid de {linhas} linhas e {colunas} colunas é: {resultado}')
```

13. Reescreva a função apresentada na questão anterior, desta vez usando programação dinâmica bottom-up para melhorar sua eficiência.

```python
def caminhos_unicos_dp(linhas, colunas):
    dp = [[1] * colunas for _ in range(linhas)]
    for i in range(1, linhas):
        for j in range(1, colunas):
            dp[i][j] = dp[i - 1][j] + dp[i][j - 1]
    return dp[-1][-1]

linhas = 3
colunas = 7
resultado = caminhos_unicos_dp(linhas, colunas)
print(f'O número de caminhos mais curtos em um grid de {linhas} linhas e {colunas} colunas é: {resultado}')
```

14. Escreva uma função que recebe um array de números positivos e retorna o maior produto de quaisquer 3 números no array. Uma abordagem usando 3 laços aninhados teria complexidade O(N3), o que seria muito ineficiente. Sua função deve ter complexidade O(N log N).

```python
def maior_produto_tres(nums):
    nums.sort()
    return max(nums[-1] * nums[-2] * nums[-3], nums[0] * nums[1] * nums[-1])
array = [1, 10, 2, 6, 5, 3]
resultado = maior_produto_tres(array)
print(f'O maior produto de 3 números é: {resultado}')
```

15. Considere uma aplicação que possui um componente para seleção de cores (que será usado, por exemplo, para definir a cor de texto e elementos gráficos manipulados na aplicação). Cada cor será identificada por um nome (str) e pelos valores de vermelho, verde e azul que a definem no sistema RGB (3 valores int separados). Na implementação deste componente, as cores disponíveis serão representadas como dicionários, e armazenadas em uma lista conforme o exemplo apresentado abaixo. Desenvolva uma função que implemente o algoritmo quicksort para ordenar uma lista de cores usando os seguintes critérios: (1) os registros devem ser ordenados primariamente em ordem crescente da intensidade do vermelho (R); (2) caso haja empate em relação ao R, deve ser considerada a ordem crescente da intensidade do verde (G); (3) caso haja empate em relação a R e G, deve ser considerada a ordem crescente da intensidade do azul (B).

```python
cores = [
    {"nome": "preto", "r": 0, "g": 0, "b": 0},
    {"nome": "branco", "r": 255, "g": 255, "b": 255},
    {"nome": "vermelho", "r": 255, "g": 0, "b": 0},
    {"nome": "laranja", "r": 255, "g": 153, "b": 0},
    {"nome": "amarelo", "r": 255, "g": 255, "b": 0},
    {"nome": "verde", "r": 0, "g": 255, "b": 0},
    {"nome": "azul", "r": 0, "g": 0, "b": 255},
    {"nome": "roxo", "r": 153, "g": 0, "b": 255}
]

def quicksort_cores(cores):
    if len(cores) <= 1:
        return cores
    pivo = cores[len(cores) // 2]
    menores = [c for c in cores if (c["r"], c["g"], c["b"]) < (pivo["r"], pivo["g"], pivo["b"])]
    iguais = [c for c in cores if (c["r"], c["g"], c["b"]) == (pivo["r"], pivo["g"], pivo["b"])]
    maiores = [c for c in cores if (c["r"], c["g"], c["b"]) > (pivo["r"], pivo["g"], pivo["b"])]
    return quicksort_cores(menores) + iguais + quicksort_cores(maiores)

cores_ordenadas = quicksort_cores(cores)
for cor in cores_ordenadas:
    print(f"{cor['nome']} - RGB({cor['r']}, {cor['g']}, {cor['b']})")
```

No TP2, abordamos a implementação de funcionalidades de ordenação no contexto de uma aplicação que processa registros de candidatos em um processo seletivo, representados conforme o exemplo apresentado abaixo. Retomando o contexto daquela aplicação, vamos agora abordar uma funcionalidade para identificação dos candidatos selecionados para as vagas disponíveis. Considerando que para o cargo alvo de um determinado processo seletivo há K vagas (K pode variar de um processo para outro), desenvolva uma função que implemente o algoritmo quickselect para encontrar (em uma lista não ordenada com mesmo formato do exemplo abaixo) e retornar os K primeiros colocados no processo considerando os seguintes critérios: (1) serão selecionados primariamente os candidatos com pontuação mais alta; (2) caso haja empate na pontuação, serão selecionados os candidatos com maior idade.
A função deve receber a lista contendo todos os candidatos (no formato abaixo) e o nº de vagas (K), retornando outra lista contendo apenas os K candidatos selecionados de acordo com os critérios acima (a lista retornada não precisa estar ordenada).

```python
candidatos = [
    {"ni": 1, "nome": "José", "idade": 28, "pontos": 97},
    {"ni": 2, "nome": "Márcia", "idade": 36, "pontos": 95},
    {"ni": 3, "nome": "André", "idade": 22, "pontos": 90},
    {"ni": 4, "nome": "Bruna", "idade": 38, "pontos": 95}
]

import random

def quickselect_candidatos(candidatos, k):
    if not candidatos:
        return []
    pivo = random.choice(candidatos)  # Escolhe um pivô aleatório
    maiores = [c for c in candidatos if (c["pontos"], c["idade"]) > (pivo["pontos"], pivo["idade"])]
    menores = [c for c in candidatos if (c["pontos"], c["idade"]) < (pivo["pontos"], pivo["idade"])]
    iguais = [c for c in candidatos if (c["pontos"], c["idade"]) == (pivo["pontos"], pivo["idade"])]
    
    if len(maiores) >= k:
        return quickselect_candidatos(maiores, k)
    if len(maiores) + len(iguais) >= k:
        return maiores + iguais[:k - len(maiores)]
    return maiores + iguais + quickselect_candidatos(menores, k - len(maiores) - len(iguais))

k = 3
melhores_candidatos = quickselect_candidatos(candidatos, k)

print(f"Os {k} melhores candidatos são:")
for candidato in melhores_candidatos:
    print(f"Nome: {candidato['nome']}, Idade: {candidato['idade']}, Pontos: {candidato['pontos']}")
```
