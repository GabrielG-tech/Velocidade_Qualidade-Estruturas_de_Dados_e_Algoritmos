2) Pegue um baralho de cartas: retire as 13 cartas de espadas, reserve o resto e embaralhe as cartas de espadas. Elabore um algoritmo para ordená-las por número sob as seguintes restrições

a. Todas as cartas devem ser seguradas em uma mão. Esta é a "primeira" mão.

b. Inicialmente, as cartas embaralhadas estão todas empilhadas com as faces em uma direção, de modo que apenas uma carta seja visível.

c. Inicialmente, todas as cartas são seguradas entre o polegar e o indicador da primeira mão.

d. A carta visível na pilha pode ser retirada usando a outra mão e colocada entre quaisquer dedos da primeira mão. Ela só pode ser colocada na frente ou atrás das cartas na pilha de cartas entre esses dedos.

e. A outra mão pode segurar apenas uma carta por vez e deve colocá-la em algum lugar na primeira mão antes de pegar outra carta visível de uma das pilhas.

f. O algoritmo termina quando todas as cartas estão ordenadas em uma única pilha na mão.

*Algoritmo para ordenar as cartas de Espadas:*

- Pegue a pilha de cartas embaralhadas na primeira mão.

- Percorra a pilha várias vezes, comparando cartas consecutivas.

- Para cada par de cartas adjacentes:
    - Retire a primeira carta visível com a outra mão.
    - Retire a próxima carta da pilha e compare.
    - Se a primeira carta for maior que a segunda, troque a ordem ao recolocá-las na primeira mão.
    - Caso contrário, recoloque-as na mesma ordem.

- Continue passando pelas cartas repetidamente até que nenhuma troca seja necessária.

- O algoritmo termina quando todas as cartas estão ordenadas.

5) A seguinte função encontra o maior número único dentro de um array, mas tem uma eficiência de O(N²). Reescreva a função para que se torne uma eficiência O(N):

![alt text](<TP1 - 5.png>)

```python
array = [3, 7, 2, 9, 5, 10, 1, 8]

def greatest_number(array):
    if not array:  # Verifica se o array está vazio
        return None  

    max_val = array[0]  # Assume o primeiro elemento como o maior
    for num in array:   # Percorre o array uma única vez
        if num > max_val:
            max_val = num  # Atualiza o maior número encontrado
    
    return max_val

maior_numero = greatest_number(array)
print("O maior número é:", maior_numero)
```

6) Imagine que você tem um tabuleiro de xadrez e coloca um único grão de arroz em um quadrado. No segundo quadrado, você coloca 2 grãos de arroz, já que isso é o dobro da quantidade de arroz no quadrado anterior. No terceiro quadrado, você coloca 4 grãos. No quarto quadrado, você coloca 8 grãos, e no quinto quadrado, você coloca 16 grãos, e assim por diante.

a. Escreva a função em python que calcula em qual quadrado do tabuleiro você precisará colocar uma determinada quantidade de grãos de arroz. Por exemplo, para 16 grãos, a função retornará 5, já que você colocará os 16 grãos no quinto quadrado.

```python
def encontrar_quadrado(n):
    if n < 1:
        return "Quantidade inválida"
    
    quadrado = 1
    graos = 1
    
    while graos < n:
        quadrado += 1
        graos *= 2
    
    return quadrado

# Exemplos:
n = 8
print(f"Para {n} grãos será necessário por no {encontrar_quadrado(n)}º quadrado")

n = 16
print(f"Para {n} grãos será necessário por no {encontrar_quadrado(n)}º quadrado")
```

b. Use a Notação Big O para descrever a complexidade de tempo da função que você acabou de criar.

- A quantidade de grãos no quadrado segue uma progressão geométrica com razão 2: (1, 2, 4, 8, 16, ...).
- A cada iteração do `while`, o número de grãos dobra (graos = graos*2).
- O loop continua até que o número de grãos seja maior ou igual a `n`.
- Como dobrar um número `k` vezes resulta em (2^k), precisamos resolver a equação 2^k >= n.
- Isso leva a k = log(n), o que significa que a função executa aproximadamente **log₂(n)** iterações.

Logo a complexidade de tempo da função `encontrar_quadrado(n)` é **O(log n)**.

O índice interno sempre vai da esquerda para a direita, encontrando o maior item e levando-o para a direita. Modifique o método bubbleSort() para que ele seja bidirecional.
